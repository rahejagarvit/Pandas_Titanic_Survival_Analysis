## Analysis of Titanic Survival Rates
The Titanic disaster is one of the most well-known maritime tragedies in history. This project explores the classic dataset of Titanic passengers to uncover factors that influenced a passenger's likelihood of survival. By cleaning, exploring, and analyzing the data, we aim to answer questions such as: Did gender, wealth, or family size play a role in who survived?

This analysis uses the pandas library for data manipulation and seaborn for loading the dataset.

```python
import pandas as pd
import seaborn as sns
```

## Data Loading and Initial Inspection
At first, we load the dataset directly from the Seaborn library and perform an initial inspection to understand its structure, data types, and identify any immediate issues like missing values.

### Loading the Dataset

```python
df = sns.load_dataset('titanic')
```

### Viewing the First Few Rows of the Dataset

```python
print('The first 10 rows of the Dataset are as follows:')
df.head(10)
```

### What is the size of the Dataset?

```python
print('The size of the dataset is :',df.shape)
```

### What type of Data Types are Present in the Dataset?

```python
print('The dataset has the following Data Types : ') 
df.info()
```

## Checking for Null Values

```python
print('Checking for Null Values in Different Columns :')
df.isnull().sum()
```

## Observations

The age column has 177 missing values. We will fill these with the median age, which is more robust to outliers than the mean.

The embarked and embark_town columns have 2 missing values. We will fill them with the mode (the most common port).

The deck column is missing too many values (688 out of 891) to be useful, so we will drop it.

The data type of Sex Column should be changed to catgeory instead of Object.

### Substituting the missing 'age' values with the median age

```python
median_age = df['age'].median()
df['age'] = df['age'].fillna(median_age)
```

### Then we fill the missing 'embarked' and 'embark_town' values with the mode

```python
mode_embarked = df['embarked'].mode()[0]
mode_embark_town = df['embark_town'].mode()[0]
df['embarked'] = df['embarked'].fillna(mode_embarked)
df['embark_town'] = df['embark_town'].fillna(mode_embark_town)
```

### Since the Deck Column has 688 missing values out of 891 we Drop It

```python
df.drop(columns=['deck'], inplace = True)
```

### Now we convert the 'sex' column to from Object to Category type for efficiency

```python
df['sex'] = df['sex'].astype('category')
```

### Verifying if all the missing values are  now gone

```python
df.isnull().sum()
```

### Now we obtain some insights from the Dataset

To better analyze survival patterns, we can create new columns (features) from the existing data. We will create columns to determine a passenger's family size and whether they were traveling alone.

### Addition of family_size column

```python
df['family_size'] = df['sibsp'] + df['parch'] + 1
```

### Addition of a Column In order to check if the person is living alone

```python
df['is_alone'] = (df['family_size'] == 1)
```

### Now we see the first few rows of the dataset in order to see the newly added rows

```python
df.head(8)
```

### Exploratory Data Analysis (EDA)

Now that our data is clean, we can answer our key questions by grouping and aggregating the data.

### Q1: What was the overall survival rate?

```python
survival_rate = df['survived'].mean()
```

```python
print(f'The overall survival rate was : {survival_rate:.2%}')
```

### Q2: Was the overall survival rate influenced by Gender?

```python
gen_surv_rate = df.groupby('sex', observed = True)['survived'].mean()
```

```python
gen_surv_rate.to_frame().style.format('{:.2%}')
```

Observation - Female passengers had a much higher survival rate (74%) compared to male passengers (19%).

### Q3: Did the class in which passengers travelled influence survival?

```python
class_surv = df.groupby('pclass')['survived'].mean()
```

```python
class_surv.to_frame().style.format('{:.2%}')
```

Observation - There is a clear correlation between wealth and survival. First-class passengers had the highest survival rate (63%), while third-class passengers had the lowest (24%).

### Q4. Was survival of an Individual Impacted by travelling alone?

```python
alone_surv = df.groupby('is_alone')['survived'].mean()
```

```python
alone_surv.to_frame().style.format('{:.2%}')
```

Observation - Passengers who were with family had a higher survival rate (50.6%) than those who were traveling alone (30.4%).

### Conclusion

This analysis of the Titanic dataset reveals several clear factors that influenced survival. The data strongly suggests that females, passengers in higher classes, and those traveling with family had a significantly greater chance of surviving the disaster.
